#MERGE SELURUH FILES
Agar analisis menjadi lebih mudah maka perlu untuk menggabungkan data menjadi satu

```python
import glob

# Ambil semua file checkpoint dari folder processed
files = glob.glob("C:\\Users\\asus3\\Documents\\CPSTNPROJECT\\capstone-ds\\Data\\processed\\linkedin_jobs_*.csv")

# Merge semua jadi satu
df = pd.concat([pd.read_csv(f) for f in files], ignore_index=True)
df.to_csv("C:\\Users\\asus3\\Documents\\CPSTNPROJECT\\capstone-ds\\Data_Wrangling\\Data_Fixed\\Data_Merged.csv", index=False)

print(f"Total baris sebelum cleaning: {len(df)}")
```

```python
print("Jumlah duplikasi: ", df.duplicated().sum())
```

```python
df = df.drop_duplicates(subset=["title", "company", "location", "job_description"])
df = df.dropna(subset=["title", "company", "location", "job_description"])
df = df[df["job_description"].str.len() >= 200]
df.to_csv("C:\\Users\\asus3\\Documents\\CPSTNPROJECT\\capstone-ds\\Data_Wrangling\\Data_Fixed\\linkedin_jobs_clean.csv", index=False)
```
